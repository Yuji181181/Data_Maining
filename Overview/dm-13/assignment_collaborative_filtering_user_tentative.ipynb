#### Return Series of similiarity of users  
by using Pearson correlation coefficient (-1 .. 1).  

```python
def get_sim_ser_by_pearson(df_users, ser_target):
    corr_all = df_users.corrwith(ser_target, axis=1).dropna()
    return corr_all
```
