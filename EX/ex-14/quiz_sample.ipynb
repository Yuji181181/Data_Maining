```python
import numpy as np
import pandas as pd
```

```python
csv_in = 'item.csv'
# min number of common items between target user and users in DB
min_common_users = 4
min_common_items = 4

csv_out = 'item_sim-min{}.csv'.format(min_common_users)

# To show more rows and columns
pd.options.display.max_rows = 999 
pd.options.display.max_columns = 999 
```

```python
df = pd.read_csv(csv_in, sep=',', skiprows=0, header=0)
print(df.shape)
display(df.info())
display(df.head())
```

```python
df_corr = df.corr(min_periods=min_common_items)
display(df_corr.head())
```

```python
df_corr.to_csv(csv_out, index=False)
```
