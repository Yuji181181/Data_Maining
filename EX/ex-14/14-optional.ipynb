```python
import numpy as np
import pandas as pd
import pyfpgrowth
```

```python
csv_in = 'market_ex.csv'
```

```python
df = pd.read_csv(csv_in, sep=',', skiprows=0, header=None)
print(df.shape)
print(df.info())
display(df.head())
```

```python
df = df.replace(np.nan, '00nan')
display(df.head())
```

```python
ser_items = pd.Series(df.values.flatten())
top_items = ser_items.value_counts()
print(top_items.head())
```

```python
item1 = top_items.index[1]
item2 = top_items.index[2]
print(item1, item2)
```

```python
df_bool1 = (df == item1)
display(df_bool1.head())
n_sc1 = df_bool1.values.sum()
print(n_sc1)
```

```python
df_bool2 = (df == item2)
display(df_bool2.head())
n_sc2 = df_bool2.values.sum()
print(n_sc2)
```

```python
df_bool3 = ((df == item1) | (df == item2))
display(df_bool3.head())
count = df_bool3.values.sum(axis=1)
print(count[:20])
```

```python
n_either = np.sum(count>0)
print(n_either)
```

```python
n_both = np.sum(count==2)
print(n_both)
```

```python
n_all = df.shape[0]
```

```python
print('Jac(sc1 => sc2):',  n_both / n_either)
```

```python
print('Supp(sc1 => sc2):', n_both / n_all)
```

```python
print('Conf(sc1 => sc2):', n_both / n_sc1)  
print('Conf(sc2 => sc1):', n_both / n_sc2) 
```

```python
a = n_both / n_sc1
b = n_sc2 / n_all
print('Lift(sc1 => sc2):', a/b)
```
