```python
df=pd.read_csv('market.csv')

print(df.shape)
print(df.info())
display(df.head())
```

```python
df_id = pd.read_csv('ID_data.csv')

print(df_id.shape)
print(df_id.info())
display(df_id.head())
```

```python
df_merge=pd.merge(df,df_id,on='顧客ID', how='left')
```

```python
display(df_merge.head())
```

```python
df_agri = df_merge[df_merge['大カテゴリ名']=='農産']
```

```python
df_agri_pivot = pd.pivot_table(df_agri, index='顧客ID', values="個数", columns="小カテゴリ名", aggfunc="sum")
```
